# Pretrained model

```python
import timm
import torch
import torchvision.transforms as T
from PIL import Image
from urllib.request import urlopen
```

```python
model = timm.create_model("hf-hub:BVRA/beit_base_patch16_224.in1k_ft_fungitastic_224", pretrained=True)
model = model.eval()

train_transforms = T.Compose([T.Resize((224, 224)),
                              T.ToTensor(),
                              T.Normalize([0.5, 0.5, 0.5], [0.5, 0.5, 0.5])])

img = Image.open("/content/drive/MyDrive/Colab Notebooks/lewagon/Kinoko/data/image_dataset/poisonous/Helvella_crispa/Helvella_crispa25.png")
#img = Image.open("data/image_dataset/poisonous/Russula_emetica/Russula_emetica25.png")

output = model(train_transforms(img).unsqueeze(0))
output
```

```python
import torch.nn.functional as F
# Get probabilities
probs = F.softmax(output, dim=1)
top5 = probs.topk(5)

# Get class labels from the model config
print(model.pretrained_cfg)  # check if label map is embedded

# Or access directly:
label_map = model.pretrained_cfg.get('label_map') or model.pretrained_cfg.get('classes')

# Top predictions
for prob, idx in zip(top5.values[0], top5.indices[0]):
    label = label_map[idx.item()] if label_map else idx.item()
    print(f"{label}: {prob.item():.4f}")
```

## Fine tuning

```python
import os
os.environ["KERAS_BACKEND"] = "tensorflow"

import keras
import keras_hub
import tensorflow as tf
import numpy as np
```

```python
# --- CONFIGURATION ---
# DINOv2 Base preset expects 518x518
IMG_SIZE = 518
BATCH_SIZE = 32
EPOCHS = 50
SEED = 42
```

```python
# --- DATA PREPARATION ---

train_ds = tf.keras.utils.image_dataset_from_directory(
    image_data_dir,
    labels="inferred",
    label_mode="binary",
    validation_split=0.3,
    subset="training",
    seed=SEED,
    image_size=(IMG_SIZE, IMG_SIZE), # Changed to 518
    batch_size=BATCH_SIZE
)

test_val_ds = tf.keras.utils.image_dataset_from_directory(
    image_data_dir,
    labels="inferred",
    label_mode="binary",
    validation_split=0.3,
    subset="validation",
    seed=SEED,
    image_size=(IMG_SIZE, IMG_SIZE), # Changed to 518
    batch_size=BATCH_SIZE
)

# Split test and val data
half_test_val_size = int(len(test_val_ds)/2)
test_ds = test_val_ds.take(half_test_val_size)
val_ds = test_val_ds.skip(half_test_val_size)

# Prefect for faster loading in gpu
train_ds.prefetch(tf.data.AUTOTUNE)
test_ds.prefetch(tf.data.AUTOTUNE)
val_ds.prefetch(tf.data.AUTOTUNE)
```

```python
# --- MODEL BUILDING ---

backbone = keras_hub.models.DINOV2Backbone.from_preset("dinov2_base")

# Freeze the backbone
backbone.trainable = False

# Build the model
inputs = keras.Input(shape=(IMG_SIZE, IMG_SIZE, 3))

# ImageNet Normalization
x = keras.layers.Rescaling(1/255.0)(inputs)
x = keras.layers.Normalization(
    mean=[0.485, 0.456, 0.406],
    variance=[0.229**2, 0.224**2, 0.225**2]
)(x)

# Pass to backbone
#backbone_out = backbone({"images": x})
#outputs = backbone(backbone_out)[:, 0, :]
backbone_out = backbone({"images": x})  # ← dict
outputs = backbone_out[:, 0, :]  # CLS token
```

```python
# --- CUSTOM HEAD ---
x = keras.layers.Dense(512, activation="relu")(outputs)
# x = keras.layers.BatchNormalization()(x)
x = keras.layers.Dropout(0.3)(x)
# x = keras.layers.Dense(256, activation="relu")(x)
predictions = keras.layers.Dense(1, activation="sigmoid")(x)

model = keras.Model(inputs=inputs, outputs=predictions)

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-4),
    loss="binary_crossentropy",
    metrics=["accuracy", "recall", "precision"]
)

model.summary()
```

```python
# --- CALLBACKS ---
callbacks = [
    # Stop early if val_loss stops improving
    keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    ),

    # Save the best checkpoint
    keras.callbacks.ModelCheckpoint(
        filepath="best_model.keras",
        monitor="val_loss",
        save_best_only=True
    ),

    # Reduce LR on plateau (helps squeeze out last gains)
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=3,
        min_lr=1e-7
    ),

    keras.callbacks.ModelCheckpoint(
      filepath="/content/drive/MyDrive/Colab Notebooks/lewagon/Kinoko/checkpoints/dinov2_epoch_{epoch:02d}.keras",
      save_best_only=True,
      save_freq="epoch",  # every N epochs, or "epoch" for every one
    ),
#    keras.callbacks.ModelCheckpoint(
#        filepath="gs://bucket_of_mushrooms/checkpoints/dinov2_epoch_{epoch:02d}.keras",
#        save_best_only=True,
#        save_freq=3,
#    )
]
```

```python
print(tf.config.list_physical_devices('GPU'))
```

```python
# --- TRAINING ---
print("\nStarting training...")
model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=callbacks
)
```
