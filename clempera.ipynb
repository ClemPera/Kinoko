# Pretrained model

```python
import timm
import torch
import torchvision.transforms as T
from PIL import Image
from urllib.request import urlopen
```

```python
model = timm.create_model("hf-hub:BVRA/beit_base_patch16_224.in1k_ft_fungitastic_224", pretrained=True)
model = model.eval()

train_transforms = T.Compose([T.Resize((224, 224)), 
                              T.ToTensor(), 
                              T.Normalize([0.5, 0.5, 0.5], [0.5, 0.5, 0.5])]) 

img = Image.open("data/image_dataset/poisonous/Russula_emetica/Russula_emetica25.png")

output = model(train_transforms(img).unsqueeze(0))
output
```

```python
import torch.nn.functional as F
# Get probabilities
probs = F.softmax(output, dim=1)
top5 = probs.topk(5)

# Get class labels from the model config
print(model.pretrained_cfg)  # check if label map is embedded

# Or access directly:
label_map = model.pretrained_cfg.get('label_map') or model.pretrained_cfg.get('classes')

# Top predictions
for prob, idx in zip(top5.values[0], top5.indices[0]):
    label = label_map[idx.item()] if label_map else idx.item()
    print(f"{label}: {prob.item():.4f}")
```

## Fine tuning

```python
import os
os.environ["KERAS_BACKEND"] = "tensorflow"

import keras
import keras_hub
import tensorflow as tf
import numpy as np
```

```python
# --- CONFIGURATION ---
# DINOv2 Base preset expects 518x518
IMG_SIZE = 518  
# NUM_CLASSES = 10 
BATCH_SIZE = 32
EPOCHS = 5
SEED = 42
```

```python
# --- 1. DATA PREPARATION ---

train_ds = tf.keras.utils.image_dataset_from_directory(
    image_data_dir,
    labels="inferred",
    label_mode="categorical", # Changed from 'binary' to match 10 classes
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=(IMG_SIZE, IMG_SIZE), # Changed to 518
    batch_size=BATCH_SIZE
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    image_data_dir,
    labels="inferred",
    label_mode="categorical", # Changed from 'binary'
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=(IMG_SIZE, IMG_SIZE), # Changed to 518
    batch_size=BATCH_SIZE
)
```

```python
# --- 2. MODEL BUILDING ---

backbone = keras_hub.models.DINOV2Backbone.from_preset("dinov2_base")

# Freeze the backbone
backbone.trainable = False

# Build the model
inputs = keras.Input(shape=(IMG_SIZE, IMG_SIZE, 3))

# ImageNet Normalization
x = keras.layers.Rescaling(1/255.0)(inputs)
x = keras.layers.Normalization(
    mean=[0.485, 0.456, 0.406], 
    variance=[0.229**2, 0.224**2, 0.225**2]
)(x)

# Pass to backbone
outputs = backbone(x)[:, 0, :]
```

```python
# --- 3. CUSTOM HEAD ---
x = keras.layers.Dense(512, activation="relu")(outputs)
# x = keras.layers.BatchNormalization()(x)
# x = keras.layers.Dropout(0.3)(x)
# x = keras.layers.Dense(256, activation="relu")(x)
predictions = keras.layers.Dense(2, activation="softmax")(x)

model = keras.Model(inputs=inputs, outputs=predictions)

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-4),
    loss="categorical_crossentropy",
    metrics=["accuracy", "recall", "precision"]
)

model.summary()
```

```python
# --- 4. TRAINING ---
print("\nStarting training...")
model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=EPOCHS
)
```
